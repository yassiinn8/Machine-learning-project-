```python
column_names = ['buying_price','maintenance_price','no_doors','persons_cap','size_of_luggage_boot','safety','car_ratings']

df = pd.read_csv('/content/car_evaluation.csv', header=None, names=column_names)

df.head()
```

```python
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    if col != 'car_ratings':
        plt.figure(figsize=(10, 6))
        sns.countplot(data=df, x=col, hue='car_ratings')
        plt.title(f'Relationship between '+col+' and Car Ratings')
        plt.xlabel(col)
        plt.ylabel('Count')
        plt.xticks(rotation=45)
        plt.show()
```

```python
categorical_cols = df.select_dtypes(include='object')
for col in categorical_cols:
    df[col] = LabelEncoder().fit_transform(df[col])

df.head()
```

```python
df.plot(kind='box', subplots=True, layout=(3,3), figsize=(15,7))
plt.show()
```

```python
plt.figure(figsize=(15, 8))
sns.heatmap(df.corr(numeric_only = True), annot=True)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

X = df.drop("car_ratings", axis=1)
y = df["car_ratings"]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.33, random_state=42)

model = RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced')

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```
